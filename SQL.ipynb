# SQL Basics - Your First Queries

This notebook introduces fundamental SQL operations using a sample e-commerce dataset.

```python

_results = []
_stream = sqlutils.sql_stream("-- ============================================================\n-- CELL 1: Creating Your First Table\n-- ============================================================\n\nDROP TABLE IF EXISTS customers;\nCREATE TABLE customers (\n    customer_id     INT,\n    first_name      STRING,\n    last_name       STRING,\n    email           STRING,\n    signup_date     DATE,\n    country         STRING,\n    lifetime_value  DECIMAL(10,2)\n)")
for result in _stream:
    if result.status == "success":
        display(result.result)
        _results.append(result)
    else:
        # Save partial results for debugging before raising
        if len(_results) > 0:
            for r in _results:
                globals()[f'sql_output_56m3_{r.statement_index}'] = r.result
        raise Exception(result.error)

# All statements succeeded - save indexed vars if multiple results
if len(_results) > 1:
    for r in _results:
        globals()[f'sql_output_56m3_{r.statement_index}'] = r.result

# Always save main variable
sql_output_56m3 = _results[0].result if len(_results) == 1 else [r.result for r in _results]
```
